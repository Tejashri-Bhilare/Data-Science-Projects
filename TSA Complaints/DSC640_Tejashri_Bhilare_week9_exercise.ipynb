```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the complaints by subcategory dataset
df = pd.read_csv('C:/Users/Teju/Downloads/complaints_by_subcategory.csv')


# Create a pivot table using 'sum' to aggregate duplicate values
pivot_data = df.pivot_table(
    index='airport',
    columns='subcategory',
    values='count',
    aggfunc='sum'
).fillna(0)

# Plot the heat map using Seaborn
plt.figure(figsize=(12, 8))
sns.heatmap(pivot_data, annot=True, fmt="g", cmap="YlGnBu")
plt.title("Heatmap of TSA Complaints by Subcategory and Airport")
plt.xlabel("Complaint Subcategory")
plt.ylabel("Airport Code")
plt.show()
```

```python


# Create a bar chart of complaint counts by category
plt.figure(figsize=(12, 6))
sns.barplot(x='category', y='count', data=df, palette='viridis')
plt.title("Complaint Counts by Category")
plt.xlabel("Complaint Category")
plt.ylabel("Complaint Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Import necessary libraries
import pandas as pd
import plotly.express as px

# Load the dataset
file_path = 'C:/Users/Teju/Downloads/iata_icao.csv'  # Replace with your actual file path
df = pd.read_csv(file_path)

# Aggregate data by country_code to count airports
airport_counts = df.groupby('country_code').size().reset_index(name='airport')

# Create the choropleth map
fig = px.choropleth(airport_counts,
                    locations='country_code',
                    color='airport',
                    hover_name='country_code',
                    color_continuous_scale='Viridis',
                    title='Number of Airports per Country')

# Show the map
fig.show()
```

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Load the dataset
file_path = 'C:/Users/Teju/Downloads/complaints_by_category.csv' 
df = pd.read_csv(file_path)

# Display first few rows to understand the data structure
df.head()
```

```python
# Count of complaints by category
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='category', order=df['category'].value_counts().index, palette='viridis')
plt.title('Number of Complaints by Category', fontsize=16)
plt.xlabel('Category')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Complaint distribution by category
category_counts = df['category'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('viridis', len(category_counts)))
plt.title('Complaint Distribution by Category')
plt.show()
```
